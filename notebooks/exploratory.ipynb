# Exploratory Analysis: Factor Decay Prediction

Quick checks before running the full pipeline.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
from src.data.factor_generator import FactorTreeGenerator
from src.features.structural import extract_structural_features
```

## Generate sample factor expressions

```python
gen = FactorTreeGenerator(seed=42, max_depth=4)
pool = gen.generate_pool(n=30)
for i, expr in enumerate(pool[:10]):
    print(f'{i:2d}: {expr.to_qlib_str()}')
```

## Structural feature distribution

```python
feats = pd.DataFrame([extract_structural_features(e) for e in pool])
feats.head()
```

```python
feats['factor_type'].value_counts().plot(kind='bar')
plt.title('Heuristic factor-type distribution in synthetic pool')
plt.show()
```
